```python
import array
```

```python
my_array = array.array('i')
type(my_array)
my_array1 = array.array('i',[1,2,3,4])
print(my_array1)

# Insetion to Array
my_array1.insert(100,6)
print(my_array1)
my_array1.insert(10,5)
print(my_array1)
# insertion at the begining will be O(N)
# Space complexity O(1)



```
