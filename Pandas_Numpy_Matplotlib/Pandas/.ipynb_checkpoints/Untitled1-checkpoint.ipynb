```python
def process_sheet(df):
    date = df.iloc[0, 2]
    df_dropped_columns = df.drop(df.columns[[0, 1]], axis=1)
    df_final = df_dropped_columns.drop([0, 1, 2], axis=0).reset_index(drop=True)
    df_final.columns = df_final.iloc[0]
    df_final = df_final[1:].reset_index(drop=True)
    
    new_columns = []
    for col, prefix in zip(df_final.columns, df_final.iloc[0]):
        if pd.isna(prefix):
            new_columns.append(col)
        else:
            new_columns.append(f'{col} {prefix}')
    
    df_final.columns = new_columns
    df_final = df_final.drop(df_final.index[0]).reset_index(drop=True)
    df_final['Date'] = date
    
    replacements = ['HCFA\'S NORMAL', 'APPEALS NORMAL', 'POC / STATEMENTS / LETTERS NORMAL']
    new_columns = []
    replacement_index = 0
    
    for col in df_final.columns:
        if col == 'nan NORMAL':
            new_columns.append(replacements[replacement_index])
            replacement_index += 1
        else:
            new_columns.append(col)
    
    df_final.columns = new_columns
    df_final = df_final.fillna('')
    
    return df_final

# Read the Excel file
excel_file = 'Daily Printing Sign off - Jan to April.xlsx'
all_sheets = pd.read_excel(excel_file, sheet_name=None)

# Process each sheet and concatenate the results
processed_sheets = [process_sheet(df).reset_index(drop=True) for df in all_sheets.values() if df is not None]
result_df = pd.concat(processed_sheets,axis=0, ignore_index=True)

print(result_df)
```
