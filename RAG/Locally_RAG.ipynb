```python
from PyPDF2 import PdfReader

def extract_text_from_pdf(pdf_path):
    reader = PdfReader(pdf_path)
    text = ""
    for page in reader.pages:
        text += page.extract_text()
    return text

text = extract_text_from_pdf('hands_on_ml_book.pdf')
```

```python
from faiss import IndexFlatL2
import numpy as np
from sentence_transformers import SentenceTransformer

# Initialize the model for embeddings
model = SentenceTransformer('all-MiniLM-L6-v2')

# Split text into chunks
def split_text(text, chunk_size=1000):
    return [text[i:i + chunk_size] for i in range(0, len(text), chunk_size)]

chunks = split_text(text)
embeddings = model.encode(chunks)

# Create a FAISS index
index = IndexFlatL2(embeddings.shape[1])
index.add(np.array(embeddings, dtype=np.float32))
```

```python
import subprocess
import numpy as np
from sentence_transformers import SentenceTransformer
from faiss import IndexFlatL2

# Function to run the Ollama model via command line
def run_ollama_model(prompt):
    result = subprocess.run(
        ['ollama', 'run', 'llama3.1'],
        input=prompt,
        text=True,
        capture_output=True
    )
    return result.stdout

def retrieve_relevant_chunks(query, model, index, chunks, top_k=5):
    query_embedding = model.encode([query])
    distances, indices = index.search(np.array(query_embedding, dtype=np.float32), top_k)
    relevant_chunks = [chunks[i] for i in indices[0]]
    return relevant_chunks

# Function to generate a response using retrieved chunks and Ollama
def generate_response(query, relevant_chunks):
    input_text = f"Context:\n{' '.join(relevant_chunks)}\n\nQuestion: {query}\nAnswer:"
    response = run_ollama_model(input_text)
    return response

# High-level function to answer a query
def answer_query(query, model, index, chunks):
    relevant_chunks = retrieve_relevant_chunks(query, model, index, chunks)
    response = generate_response(query, relevant_chunks)
    return response


```
