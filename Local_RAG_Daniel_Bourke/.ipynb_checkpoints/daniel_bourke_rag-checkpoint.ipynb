```python
import os 
import requests

pdf_path = "human-nutrition-text.pdf"

if not os.path.exists(pdf_path):
    print(f"[INFO] File doesn't exist, downloading....")

    url = "https://pressbooks.oer.hawaii.edu/humannutrition2/open/download?type=pdf"
    filename = pdf_path
    response = requests.get(url)
    #check if the request was successful
    if response.status_code == 200:
        with open(filename, "wb") as file:
            file.write(response.content)
        print(f"[INFO] the file has been downloaded and saved as {filename}")
    else:
        print(f"[INFO] Failed to download the file. status code: {response.status_code}")
else:
    print(f"File {pdf_path} already exists")
```

```python
import fitz
from tqdm.auto import tqdm

def text_formatter(text: str) -> str:
    """ Perfroms minor formatting on text."""
    cleaned_text = text.replace("\n"," ").strip()
    return cleaned_text

def open_and_read_pdf(pdf_path: str) -> list[disct]:
    doc = fitz.open(pdf_path)
    pages_and_texts = []
    for page_number, page in tqdm(enumerate(doc)):
        text = page.get_text()
        text = text_formatter(text=text)
        pages_and_texts.append("page")
```

```python
from PIL import Image, ImageDraw, ImageFont
import os

def create_text_image(text, image_path="text_image.png", font_size=80, padding=10, bottom_padding=20):
    # Try to load Arial font, or use a default font if Arial is not available
    try:
        # Adjust the font path to where Arial.ttf is located
        font_path = "/usr/share/fonts/truetype/msttcorefonts/Arial.ttf"
        font = ImageFont.truetype(font_path, font_size)
    except IOError:
        print("Arial font not found, using default PIL font.")
        font = ImageFont.load_default()
    
    # Create a temporary image to calculate text size
    img_temp = Image.new("RGB", (1, 1))
    draw = ImageDraw.Draw(img_temp)
    
    # Measure text size and add padding
    text_bbox = draw.textbbox((0, 0), text, font=font)
    text_width = text_bbox[2] - text_bbox[0] + 2 * padding
    text_height = text_bbox[3] - text_bbox[1] + padding + bottom_padding

    # Create a new image with calculated size based on text dimensions
    img = Image.new('RGB', (text_width, text_height), color='white')
    draw = ImageDraw.Draw(img)
    
    # Position the text with padding
    position = (padding, padding)
    
    # Add text to the image
    draw.text(position, text, fill="black", font=font)
    
    # Save and display the image
    img.save(image_path)
    display(img)  # Display the image in Jupyter Notebook
    print(f"Image with text '{text}' created and saved at {image_path}")

# Call the function to create the image with bottom padding
create_text_image("PT004", font_size=80)
```

```python
import requests
import logging
import re
from html import unescape
from bs4 import BeautifulSoup
 
 
def get_access_token():
    url = "https://login.microsoftonline.com/d331aa89-4469-4d3a-aa23-21ad6b319300/oauth2/v2.0/token"
    payload = {
        'grant_type': 'password',
        'client_id': '2e0fe7a9-2ad9-4a9f-ba24-a22c486412c8',
        'client_secret': 'kWe8Q~HfZrD_CPk2tl6mxzOtkDG8TQ7Vf8pRaa-c',
        'username': 'ds1@carecloud.com',
        'password': 'Col48997',
        'scope': 'https://graph.microsoft.com/mail.readwrite'
    }
    response = requests.post(url, data=payload)
    response.raise_for_status()
    return response.json()["access_token"]
 
 
def clean_email_body(email_body):
    """
    Cleans the email body to remove unnecessary HTML tags and escape characters.
    """
    logging.info("Cleaning email body content...")
    email_body = unescape(email_body)
    # Parse HTML and extract text content
    soup = BeautifulSoup(email_body, "html.parser")
    cleaned_body = soup.get_text(separator=" ").strip()
    logging.debug(f"Cleaned email body: {cleaned_body}")
    return cleaned_body
 
 
def extract_otp_from_email_body(email_body):
    """
    Extract the OTP from the email body using regex.
    """
    logging.info("Extracting OTP from email body...")
    cleaned_body = clean_email_body(email_body)
 
    otp_match = re.search(r"Your\s+requested\s+security\s+code\s+for\s+NGSConnex\s+is\s+(\d{4})", cleaned_body, re.IGNORECASE)
    if otp_match:
        otp = otp_match.group(1)
        logging.info(f"Extracted OTP: {otp}")
        return otp
    else:
        logging.warning("No OTP found in email body.")
        return None
 


def get_otp_from_email(sender_email):
    """
    Retrieves the OTP from the most recent email from the specified sender.
    """
    GRAPH_API_ENDPOINT = 'https://graph.microsoft.com/v1.0/'
    access_token = get_access_token()
    print(access_token)
 
    headers = {
        'Authorization': f'Bearer {access_token}',
        'Content-Type': 'application/json'
    }
 
    # Fetch the most recent email from the specified sender
    messages_response = requests.get(
        GRAPH_API_ENDPOINT + 'users/93d47ccc-0263-4b0c-93ae-2d07174c293e/mailFolders/AAMkADVjNjdhNWJkLTliNTMtNDZhMS1iNGExLTIyZDM4NzVhOWI1MgAuAAAAAACIRRulh342SZ8IJDryLOI1AQBzXztBssuzRqEpdbJvOH8qAAApgu8kAAA=/messages',
        headers=headers,
        params={
            '$filter': f"from/emailAddress/address eq '{sender_email}'",
            '$orderby': 'receivedDateTime desc',  # Sort by most recent first
            '$top': 1  # Limit to the most recent email
        }
    )
    messages_response.raise_for_status()
    messages = messages_response.json().get('value', [])
 
    if not messages:
        logging.error(f"No emails found from {sender_email}.")
        return None
 
    # Get the email body content
    message_id = messages[0]['id']
    email_content_url = f"{GRAPH_API_ENDPOINT}/users/93d47ccc-0263-4b0c-93ae-2d07174c293e/messages/{message_id}"
    email_content_response = requests.get(email_content_url, headers=headers)
    email_content_response.raise_for_status()
 
    email_body = email_content_response.json().get('body', {}).get('content', '')
 
    return extract_otp_from_email_body(email_body)

 
 
 
# Example usage
if __name__ == "__main__":
    logging.basicConfig(level=logging.INFO)
    sender_email = "muhammadjaved12@carecloud.com"
    otp = get_otp_from_email(sender_email)
 
    if otp:
        print(f"Extracted OTP: {otp}")
    else:
        print("No OTP found.")
```
