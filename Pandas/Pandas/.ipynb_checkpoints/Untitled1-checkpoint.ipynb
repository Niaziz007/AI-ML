```python
from base64 import b64decode
from Crypto.Cipher import AES
from Crypto.Protocol.KDF import PBKDF2
import hashlib

def decrypt(cipher_text, pass_phrase, salt_value, hash_algorithm, password_iterations, init_vector, key_size):
    try:
        # Convert base64 string to bytes
        cipher_text_bytes = b64decode(cipher_text)
        
        # Convert strings to bytes
        init_vector_bytes = init_vector.encode('ascii')
        salt_value_bytes = salt_value.encode('ascii')

        # Derive the key
        if hash_algorithm == "SHA1":
            hash_function = hashlib.sha1
        elif hash_algorithm == "SHA256":
            hash_function = hashlib.sha256
        else:
            raise ValueError("Unsupported hash algorithm")

        key = PBKDF2(pass_phrase, salt_value_bytes, dkLen=key_size // 8, count=password_iterations, hmac_hash_module=hash_function)

        # Create the AES cipher
        cipher = AES.new(key, AES.MODE_CBC, init_vector_bytes)

        # Decrypt the data
        plain_text_bytes = cipher.decrypt(cipher_text_bytes)

        # Remove padding
        pad = plain_text_bytes[-1]
        plain_text_bytes = plain_text_bytes[:-pad]

        # Convert bytes to string
        plain_text = plain_text_bytes.decode('utf-8')

        return plain_text
    except Exception as ex:
        return None

# Test data
cipher_text = '3Va9YuIM+En465WQRBD+1A=='
pass_phrase = "PassPhrase"
salt_value = "SaltValue"
hash_algorithm = "SHA1"
password_iterations = 2
init_vector = "0123456789111213"
key_size = 256

decrypted_text = decrypt(cipher_text, pass_phrase, salt_value, hash_algorithm, password_iterations, init_vector, key_size)
print(decrypted_text)
```

```python
from Crypto.Cipher import AES
from Crypto.Protocol.KDF import PBKDF2
from Crypto.Hash import SHA1, SHA256
from Crypto.Util.Padding import pad, unpad
import base64

def encrypt(plain_text, pass_phrase, salt_value, hash_algorithm, password_iterations, init_vector, key_size):
    # Convert strings to bytes
    init_vector_bytes = init_vector.encode('utf-8')
    salt_value_bytes = salt_value.encode('utf-8')
    plain_text_bytes = plain_text.encode('utf-8')

    # Derive the key
    if hash_algorithm == "SHA1":
        hash_function = SHA1.new()
    elif hash_algorithm == "SHA256":
        hash_function = SHA256.new()
    else:
        raise ValueError("Unsupported hash algorithm")

    key = PBKDF2(pass_phrase, salt_value_bytes, dkLen=key_size // 8, count=password_iterations, hmac_hash_module=hash_function)

    # Create the AES cipher
    cipher = AES.new(key, AES.MODE_CBC, iv=init_vector_bytes)

    # Add padding (PKCS#7)
    padded_plain_text_bytes = pad(plain_text_bytes, AES.block_size)

    # Encrypt the data
    cipher_text_bytes = cipher.encrypt(padded_plain_text_bytes)

    # Convert to base64 string
    cipher_text = base64.b64encode(cipher_text_bytes).decode('utf-8')

    return cipher_text

def decrypt(cipher_text, pass_phrase, salt_value, hash_algorithm, password_iterations, init_vector, key_size):
    try:
        # Convert base64 string to bytes
        cipher_text_bytes = base64.b64decode(cipher_text)
        
        # Convert strings to bytes
        init_vector_bytes = init_vector.encode('utf-8')
        salt_value_bytes = salt_value.encode('utf-8')

        # Derive the key
        if hash_algorithm == "SHA1":
            hash_function = SHA1.new()
        elif hash_algorithm == "SHA256":
            hash_function = SHA256.new()
        else:
            raise ValueError("Unsupported hash algorithm")

        key = PBKDF2(pass_phrase, salt_value_bytes, dkLen=key_size // 8, count=password_iterations, hmac_hash_module=hash_function)

        # Create the AES cipher
        cipher = AES.new(key, AES.MODE_CBC, iv=init_vector_bytes)

        # Decrypt the data
        plain_text_bytes = cipher.decrypt(cipher_text_bytes)

        # Remove padding (PKCS#7)
        unpadded_plain_text_bytes = unpad(plain_text_bytes, AES.block_size)

        # Convert bytes to string
        plain_text = unpadded_plain_text_bytes.decode('utf-8')

        return plain_text
    except Exception as ex:
        print(f"Error: {ex}")
        return None

# Test data
plain_text = "hk79~@Mtbc@02@"
pass_phrase = "PassPhrase"
salt_value = "SaltValue"
hash_algorithm = "SHA1"
password_iterations = 2
init_vector = "0123456789111213"
key_size = 256

# Encrypt
cipher_text = encrypt(plain_text, pass_phrase, salt_value, hash_algorithm, password_iterations, init_vector, key_size)
print(f"Encrypted: {cipher_text}")

# Decrypt
decrypted_text = decrypt(cipher_text, pass_phrase, salt_value, hash_algorithm, password_iterations, init_vector, key_size)
print(f"Decrypted: {decrypted_text}")
```

```python
from Crypto.Cipher import AES
from Crypto.Protocol.KDF import PBKDF2
import base64
import hashlib
import struct

def pad(text):
    # Calculate the number of bytes to pad
    pad_size = AES.block_size - (len(text) % AES.block_size)
    # Pad with bytes of the pad size
    return text + (chr(pad_size) * pad_size).encode()

def encrypt(plain_text, pass_phrase, salt_value, hash_algorithm, password_iterations, init_vector, key_size):
    try:
        # Convert strings to bytes
        init_vector_bytes = init_vector.encode('utf-8')
        salt_value_bytes = salt_value.encode('utf-8')
        plain_text_bytes = plain_text.encode('utf-8')

        # Derive the key
        if hash_algorithm == "SHA1":
            hash_function = hashlib.sha1
        elif hash_algorithm == "SHA256":
            hash_function = hashlib.sha256
        else:
            raise ValueError("Unsupported hash algorithm")

        
        
        key = PBKDF2(pass_phrase, salt_value_bytes, dkLen=key_size // 8, count=password_iterations, prf=lambda p, s: hash_function(p + s).digest())
        
        cipher = AES.new(key, AES.MODE_CBC, iv=init_vector_bytes)

        # Pad the plaintext
        padded_plain_text = pad(plain_text_bytes)

        # Encrypt the data
        cipher_text_bytes = cipher.encrypt(padded_plain_text)

        # Convert to base64 string
        cipher_text = base64.b64encode(cipher_text_bytes).decode('utf-8')

        return cipher_text
    except Exception as ex:
        print(f"Error: {ex}")
        return None

# Test data
plain_text = "HK79~@Mtbc@02@"
pass_phrase = "PassPhrase"
salt_value = "SaltValue"
hash_algorithm = "SHA1"
password_iterations = 2
init_vector = "0123456789111213"
key_size = 256

# Encrypt
cipher_text = encrypt("", pass_phrase, salt_value, hash_algorithm, password_iterations, init_vector, key_size)
print(f"Encrypted: {cipher_text}")
```

```python
import base64
import hashlib
from Crypto.Cipher import AES, DES
from Crypto.Protocol.KDF import PBKDF2
from Crypto.Random import get_random_bytes
from Crypto.Util.Padding import pad, unpad
from datetime import datetime

class EncryptionUtil:
    # Encryption and Decryption Constants
    passPhrase = "PassPhrase"
    saltValue = "SaltValue"
    hashAlgorithm = hashlib.sha1
    passwordIterations = 2
    initVector = "0123456789111213"
    keySize = 256

    @staticmethod
    def encrypt_phrase(plain_text):
        return EncryptionUtil.encrypt(plain_text, EncryptionUtil.passPhrase, EncryptionUtil.saltValue,
                                      EncryptionUtil.hashAlgorithm, EncryptionUtil.passwordIterations,
                                      EncryptionUtil.initVector, EncryptionUtil.keySize)

    @staticmethod
    def decrypt_phrase(cipher_text):
        return EncryptionUtil.decrypt(cipher_text, EncryptionUtil.passPhrase, EncryptionUtil.saltValue,
                                      EncryptionUtil.hashAlgorithm, EncryptionUtil.passwordIterations,
                                      EncryptionUtil.initVector, EncryptionUtil.keySize)

    @staticmethod
    def encrypt(plain_text, pass_phrase, salt_value, hash_algorithm, password_iterations, init_vector, key_size):
        salt_bytes = salt_value.encode('utf-8')
        init_vector_bytes = init_vector.encode('utf-8')
        plain_text_bytes = plain_text.encode('utf-8')

        key = PBKDF2(pass_phrase, salt_bytes, dkLen=key_size // 8, count=password_iterations, hmac_hash_module=hash_algorithm)

        cipher = AES.new(key, AES.MODE_CBC, iv=init_vector_bytes)
        cipher_text_bytes = cipher.encrypt(pad(plain_text_bytes, AES.block_size))
        return base64.b64encode(cipher_text_bytes).decode('utf-8')

    @staticmethod
    def decrypt(cipher_text, pass_phrase, salt_value, hash_algorithm, password_iterations, init_vector, key_size):
        salt_bytes = salt_value.encode('utf-8')
        init_vector_bytes = init_vector.encode('utf-8')
        cipher_text_bytes = base64.b64decode(cipher_text)

        key = PBKDF2(pass_phrase, salt_bytes, dkLen=key_size // 8, count=password_iterations, hmac_hash_module=hash_algorithm)

        cipher = AES.new(key, AES.MODE_CBC, iv=init_vector_bytes)
        plain_text_bytes = unpad(cipher.decrypt(cipher_text_bytes), AES.block_size)
        return plain_text_bytes.decode('utf-8')

    @staticmethod
    def get_encrypted_code(clear_text):
        encryption_key = "MAKV2SPBNI99212"
        clear_bytes = clear_text.encode('utf-8')
        key_bytes = hashlib.sha256(encryption_key.encode('utf-8')).digest()
        iv = b"".join([bytes([i]) for i in [0x49, 0x76, 0x61, 0x6e, 0x20, 0x4d, 0x65, 0x64, 0x76, 0x65, 0x64, 0x65, 0x76]])

        cipher = AES.new(key_bytes, AES.MODE_CBC, iv[:16])
        cipher_text_bytes = cipher.encrypt(pad(clear_bytes, AES.block_size))
        return base64.b64encode(cipher_text_bytes).decode('utf-8').replace("+", "plus")

    @staticmethod
    def encrypt_password(password):
        sha512 = hashlib.sha512()
        sha512.update(password.encode('ascii'))
        return sha512.hexdigest().upper()

    @staticmethod
    def encrypt_us(string_to_encrypt, encryption_key):
        try:
            key = encryption_key.encode('utf-8')
            iv = bytes([0x12, 0x34, 0x56, 0x78, 0x90, 0xab, 0xcd, 0xef])

            des = DES.new(key, DES.MODE_CBC, iv)
            input_byte_array = string_to_encrypt.encode('utf-8')
            cipher_text_bytes = des.encrypt(pad(input_byte_array, DES.block_size))
            return base64.b64encode(cipher_text_bytes).decode('utf-8')
        except Exception as e:
            return str(e)

    @staticmethod
    def get_file_name_or_directory_path(complete_directory_with_file_name, return_file_name=True):
        parts = complete_directory_with_file_name.split('\\')
        if return_file_name:
            return parts[-1]
        return '\\'.join(parts[:-1])

    @staticmethod
    def generate_file_name(filename, file_extension):
        return f"{datetime.now().strftime('%d%m%Y%H%M%S%f')}_{filename}{file_extension}"

# Example usage:
enc_util = EncryptionUtil()
encrypted = enc_util.encrypt_phrase("HK79~@Mtbc@02@")
print("Encrypted:", encrypted)
decrypted = enc_util.decrypt_phrase(encrypted)
print("Decrypted:", decrypted)
```

```python
from datetime import datetime

# Provided timestamps
timestamps = [
    ("in", "2024-06-06 16:50:25"),
    ("out", "2024-06-06 21:36:37"),
    ("in", "2024-06-06 21:47:45"),
    ("out", "2024-06-07 00:52:11")
]

# Convert string timestamps to datetime objects
events = [(t[0], datetime.strptime(t[1], "%Y-%m-%d %H:%M:%S")) for t in timestamps]

# Initialize total time spent
total_time_spent = 0

# Calculate total time spent in office
for i in range(0, len(events), 2):
    check_in = events[i][1]
    check_out = events[i + 1][1]
    time_spent = check_out - check_in
    total_time_spent += time_spent.total_seconds()

# Convert total time spent from seconds to hours, minutes, and seconds
hours, remainder = divmod(total_time_spent, 3600)
minutes, seconds = divmod(remainder, 60)

print(f"Total time spent in office: {int(hours)} hours, {int(minutes)} minutes, and {int(seconds)} seconds.")
```
