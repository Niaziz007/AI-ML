```python
import os
import gradio as gr
from dotenv import load_dotenv
import openai

# 1. Load .env (for OPENAI_API_KEY)
load_dotenv()
openai.api_key = os.getenv("OPENAI_API_KEY")

# 2. Define the Gradio callback using OpenAI
def summarize(prompt: str) -> str:
    try:
        response = openai.ChatCompletion.create(
            model="gpt-4o",  # or "gpt-4" if available
            messages=[
                {"role": "system",  "content": "You are a helpful assistant that summarizes research prompts."},
                {"role": "user",    "content": prompt},
            ],
            temperature=0.5,
            max_tokens=150,
        )
        return response.choices[0].message["content"].strip()
    except Exception as e:
        return f"⚠️ Error: {e}"

# 3. Build and launch the Gradio UI
iface = gr.Interface(
    fn=summarize,
    inputs=gr.Textbox(lines=2, placeholder="Enter your research prompt here…"),
    outputs="text",
    title="📝 Research Summarizer",
    description="Type a prompt and hit **Generate** to get a summarized response.",
    allow_flagging="never",
)

if __name__ == "__main__":
    iface.launch(share=False)
```
